```python
from keras.models import Model
import keras.layers
```

```python
a = Input(shape=(28,28,1))
```

```python
hidden1 = Dense(128, activation='relu')(a)
hidden2 = Dense(64, activation='relu')(hidden1)

output1 = Dense(1, activation='sigmoid')(hidden2)
output2 = Dense(1, activation='sigmoid')(hidden2)
```

```python
model = Model(inputs=a, outputs=[output1, output2])
```

```python
model.summary()
```

```python
from keras.utils import plot_model
plot_model(model, show_shapes=True)
```

```python
input_a = Input(shape=(392,))
hidden1 = Dense(30, activation='softmax')(input_a)
hidden2 = Dense(10, activation='softmax')(hidden1)

input_b = Input(shape=(392,))
concat = keras.layers.concatenate([hidden2, input_b])
output1 = Dense(1, activation='sigmoid')(concat)
```
